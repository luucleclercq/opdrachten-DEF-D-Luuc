```python
# Voeg hier je plot in.
f_0=2.5
omega_0=5*np.pi
omega=np.linspace(0,100,1000)
A= np.abs(1/(1-(omega**2)/(omega_0**2)))

plt.loglog(omega,A)
plt.xlabel('log(omega)')
plt.ylabel('log(A)')
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.

delta_x=9.81/(omega_0**2)
print(delta_x)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().
def func(m):
    return (9.81*m)/delta_x 

m_values = np.linspace(0, 10, 100)
C_values = func(m_values)

plt.plot(m_values, C_values, label='C(m)')
plt.plot(5*(m_values),C_values, label='C(5m)')
plt.plot(2*m_values,C_values, label='C(2m)')
plt.xlabel('Massa (kg)')
plt.ylabel('Veerconstante (N/m)')
plt.savefig('opdr3.png')
plt.legend()
```
